## Part 1: Central Tendency Solutions

### Solution 1.1: E-commerce Transaction Analysis

```
# Problem 1.1 Data
transaction_amounts = [25, 30, 35, 28, 45, 50, 32, 38, 42, 29, 
                       31, 48, 36, 33, 40, 27, 350, 34, 37, 41]

# Task 1: Calculate central tendency measures
mean_amount = np.mean(transaction_amounts)
median_amount = np.median(transaction_amounts)
mode_result = stats.mode(transaction_amounts, keepdims=True)
mode_amount = mode_result.mode[0] if mode_result.count[0] > 1 else "No clear mode"

# Task 2: Print results
print("Central Tendency Analysis")
print("="*50)
print(f"Mean:   ${mean_amount:.2f}")
print(f"Median: ${median_amount:.2f}")
print(f"Mode:   {mode_amount}")
print()
print(f"Difference: Mean is ${mean_amount - median_amount:.2f} higher than median")
print()

# Task 3: Identify outliers
print("Outlier Analysis:")
print(f"The value $350 is an outlier (much larger than others)")
print(f"Without outlier: Mean would be ${np.mean([x for x in transaction_amounts if x != 350]):.2f}")
print(f"The outlier inflates the mean by ${mean_amount - np.mean([x for x in transaction_amounts if x != 350]):.2f}")
print()
print("Conclusion: Median (${:.2f}) better represents typical spending".format(median_amount))

# Task 4: Create visualization
plt.figure(figsize=(10, 6))
plt.hist(transaction_amounts, bins=15, alpha=0.7, color='skyblue', edgecolor='black')
plt.axvline(mean_amount, color='red', linestyle='--', linewidth=2, label=f'Mean: ${mean_amount:.2f}')
plt.axvline(median_amount, color='green', linestyle='--', linewidth=2, label=f'Median: ${median_amount:.2f}')
plt.xlabel('Transaction Amount ($)')
plt.ylabel('Frequency')
plt.title('E-commerce Transaction Amounts Distribution')
plt.legend()
plt.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()
```

### Solution 1.2: Regional Sales Comparison

```
# Problem 1.2 Data
sales_data = pd.DataFrame({
    'Region': ['North']*15 + ['South']*15 + ['East']*15,
    'Daily_Sales': list(np.random.normal(15000, 2000, 15)) + 
                   list(np.random.normal(18000, 3000, 15)) + 
                   list(np.random.normal(12000, 1500, 15))
})

# Task 1: Calculate statistics by region
regional_stats = sales_data.groupby('Region')['Daily_Sales'].agg(['mean', 'median', 'std']).round(2)
print("Regional Sales Statistics")
print("="*60)
print(regional_stats)
print()

# Task 2: Identify highest average
highest_region = regional_stats['mean'].idxmax()
print(f"Task 2: Highest average sales: {highest_region} (${regional_stats.loc[highest_region, 'mean']:,.2f})")
print()

# Task 3: Assess consistency
print("Task 3: Consistency Analysis (Mean vs Median difference):")
for region in regional_stats.index:
    mean_val = regional_stats.loc[region, 'mean']
    median_val = regional_stats.loc[region, 'median']
    diff = abs(mean_val - median_val)
    print(f"  {region}: Difference = ${diff:.2f} ({'Consistent' if diff < 500 else 'Some variation'})")
print()

# Task 4: Create visualization
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))

# Bar chart
regional_stats['mean'].plot(kind='bar', ax=ax1, color=['#1f77b4', '#ff7f0e', '#2ca02c'], edgecolor='black')
ax1.set_title('Average Daily Sales by Region', fontweight='bold')
ax1.set_ylabel('Average Daily Sales ($)')
ax1.set_xlabel('Region')
ax1.grid(True, alpha=0.3, axis='y')
ax1.tick_params(axis='x', rotation=0)

# Box plot
sales_data.boxplot(column='Daily_Sales', by='Region', ax=ax2, patch_artist=True)
ax2.set_title('Sales Distribution by Region', fontweight='bold')
ax2.set_ylabel('Daily Sales ($)')
ax2.get_figure().suptitle('')

plt.tight_layout()
plt.show()
```

## Part 2: Measures of Variability Solutions

### Solution 2.1: Product Quality Consistency

```
# Problem 2.1 Data
line_A = [100.2, 99.8, 100.5, 100.1, 99.9, 100.3, 100.0, 99.7, 100.4, 100.2,
          99.8, 100.1, 100.3, 99.9, 100.0, 100.2, 99.8, 100.4, 100.1, 99.9]

line_B = [98.5, 101.2, 99.3, 102.1, 97.8, 100.5, 99.1, 101.8, 98.2, 102.5,
          99.7, 100.8, 98.9, 101.5, 99.4, 100.2, 98.7, 101.9, 99.2, 100.6]

# Task 1: Calculate variability measures
def calc_variability(data, name):
    mean_val = np.mean(data)
    variance = np.var(data, ddof=1)
    std_dev = np.std(data, ddof=1)
    data_range = np.max(data) - np.min(data)
    cv = (std_dev / mean_val) * 100
    
    print(f"{name}:")
    print(f"  Mean:         {mean_val:.4f}")
    print(f"  Range:        {data_range:.4f}")
    print(f"  Variance:     {variance:.4f}")
    print(f"  Std Dev:      {std_dev:.4f}")
    print(f"  CV:           {cv:.2f}%")
    print()
    return {'mean': mean_val, 'std': std_dev, 'cv': cv}

print("Variability Analysis")
print("="*60)
stats_A = calc_variability(line_A, "Line A")
stats_B = calc_variability(line_B, "Line B")

# Task 2: Compare consistency
print("Task 2: Consistency Comparison")
print(f"Line A CV: {stats_A['cv']:.2f}%")
print(f"Line B CV: {stats_B['cv']:.2f}%")
print(f"Conclusion: Line A is more consistent (lower CV)")
print()

# Task 3: Create box plots
fig, ax = plt.subplots(figsize=(10, 6))
ax.boxplot([line_A, line_B], labels=['Line A', 'Line B'], patch_artist=True)
ax.set_ylabel('Widget Weight (grams)')
ax.set_title('Manufacturing Line Comparison - Product Consistency', fontweight='bold')
ax.grid(True, alpha=0.3, axis='y')
plt.tight_layout()
plt.show()

# Task 4: Calculate percentage within ±1 std dev
mean_A = stats_A['mean']
std_A = stats_A['std']
within_1std = sum(1 for x in line_A if mean_A - std_A <= x <= mean_A + std_A)
pct_within = (within_1std / len(line_A)) * 100
print(f"Task 4: Percentage of Line A within ±1 std dev: {pct_within:.1f}%")
print(f"        ({within_1std} out of {len(line_A)} values)")
```

### Solution 2.2: Outlier Detection Using IQR

```
# Problem 2.2 Data
purchase_amounts = [45, 52, 48, 55, 60, 58, 51, 49, 53, 47, 150, 56, 50, 54, 
                    62, 48, 52, 10, 58, 55, 51, 59, 48, 52, 200, 56, 50, 54, 61]

# Task 1: Calculate Q1, Q3, and IQR
Q1 = np.percentile(purchase_amounts, 25)
Q3 = np.percentile(purchase_amounts, 75)
IQR = Q3 - Q1

print("Outlier Detection Using IQR Method")
print("="*60)
print(f"Q1 (25th percentile): ${Q1:.2f}")
print(f"Q3 (75th percentile): ${Q3:.2f}")
print(f"IQR:                  ${IQR:.2f}")
print()

# Task 2: Calculate outlier bounds
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR
print(f"Lower bound: ${lower_bound:.2f}")
print(f"Upper bound: ${upper_bound:.2f}")
print()

# Task 3: Identify outliers
outliers = [x for x in purchase_amounts if x < lower_bound or x > upper_bound]
print(f"Task 3: Outliers detected: {outliers}")
print(f"        Number of outliers: {len(outliers)}")
print(f"        Percentage: {len(outliers)/len(purchase_amounts)*100:.1f}%")
print()

# Task 4: Create box plot
plt.figure(figsize=(10, 6))
plt.boxplot(purchase_amounts, vert=True, patch_artist=True,
            boxprops=dict(facecolor='lightblue', alpha=0.7))
plt.ylabel('Purchase Amount ($)')
plt.title('Customer Purchase Amounts - Outlier Detection', fontweight='bold')
plt.grid(True, alpha=0.3, axis='y')

# Annotate outlier bounds
plt.axhline(lower_bound, color='red', linestyle='--', alpha=0.5, label=f'Lower Bound: ${lower_bound:.2f}')
plt.axhline(upper_bound, color='red', linestyle='--', alpha=0.5, label=f'Upper Bound: ${upper_bound:.2f}')
plt.legend()
plt.tight_layout()
plt.show()
```
