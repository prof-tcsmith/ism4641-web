## 📦 Part 3: Creating Modules

In Google Colab, we create modules using the `%%writefile` magic command. This creates a Python file that can be imported.

### Module 1: coffee_calculations.py

````python
%%writefile coffee_calculations.py
"""
Coffee Calculations Module
Functions for financial calculations
"""

def calculate_order_total(items, prices, is_member=False):
    """Calculate the total for one order."""
    total = 0.0
    for item in items:
        if item in prices:
            total += prices[item]
    if is_member:
        total = total * 0.9  # 10% discount
    return total

def calculate_daily_revenue(sales_list, prices, member_list):
    """Calculate total revenue for a day's sales."""
    total_revenue = 0.0
    for customer_id, items, _ in sales_list:
        is_member = customer_id in member_list
        order_total = calculate_order_total(items, prices, is_member)
        total_revenue += order_total
    return total_revenue

def apply_discount(amount, discount_percent):
    """Apply a percentage discount to an amount."""
    return amount * (1 - discount_percent / 100)
````

### Module 2: coffee_analytics.py

````python
%%writefile coffee_analytics.py
"""
Coffee Analytics Module
Functions for analyzing sales data
"""

def find_best_sellers(sales_list, top_n=3):
    """Find the most frequently ordered items."""
    item_counts = {}
    for _, items, _ in sales_list:
        for item in items:
            if item in item_counts:
                item_counts[item] += 1
            else:
                item_counts[item] = 1
    
    items_list = [(item, count) for item, count in item_counts.items()]
    items_list.sort(key=lambda x: x[1], reverse=True)
    return items_list[:top_n]

def count_unique_customers(sales_list):
    """Count the number of unique customers."""
    customers = set()
    for customer_id, _, _ in sales_list:
        if customer_id:
            customers.add(customer_id)
    return len(customers)

def calculate_average_order(sales_list):
    """Calculate the average order value."""
    if len(sales_list) == 0:
        return 0
    total = sum(amount for _, _, amount in sales_list)
    return total / len(sales_list)
````

### Module 3: coffee_reports.py

````python
%%writefile coffee_reports.py
"""
Coffee Reports Module
Functions for generating reports
"""

from coffee_analytics import find_best_sellers

def generate_report(location_name, sales_data, **report_sections):
    """Generate a customizable sales report."""
    report = f"{'='*30}\n"
    report += f"{location_name} Sales Report\n"
    report += f"{'='*30}\n"
    
    if report_sections.get('total_revenue', False):
        total = sum(amount for _, _, amount in sales_data)
        report += f"Total Revenue: ${total:.2f}\n"
    
    if report_sections.get('customer_count', False):
        customers = set(c for c, _, _ in sales_data if c)
        report += f"Unique Customers: {len(customers)}\n"
    
    if report_sections.get('transaction_count', False):
        report += f"Total Transactions: {len(sales_data)}\n"
    
    best_sellers_count = report_sections.get('best_sellers', 0)
    if best_sellers_count > 0:
        best_items = find_best_sellers(sales_data, best_sellers_count)
        report += f"\nTop {best_sellers_count} Items:\n"
        for item, count in best_items:
            report += f"  - {item}: sold {count} times\n"
    
    return report

def compare_locations(*location_sales, **options):
    """Compare sales across multiple locations."""
    metric = options.get('metric', 'revenue')
    results = {}
    
    for i, sales_list in enumerate(location_sales, 1):
        location_name = f"location_{i}"
        
        if metric == 'transactions':
            results[location_name] = len(sales_list)
        elif metric == 'items':
            results[location_name] = sum(len(items) for _, items, _ in sales_list)
        elif metric == 'revenue':
            results[location_name] = sum(amount for _, _, amount in sales_list)
    
    return results

def format_currency(amount):
    """Format amount as currency."""
    return f"${amount:.2f}"
````

## 🎯 Main Program

Create a main program that uses all the modules together:

````python
def main():
    """
    Main program that demonstrates using all modules together.
    """
    print("☕ COZY COFFEE CORNER - DAILY ANALYTICS")
    print("=" * 50)
    
    # Calculate revenue for both locations
    westshore_rev = calc.calculate_daily_revenue(westshore_sales, menu_prices, members)
    downtown_rev = calc.calculate_daily_revenue(downtown_sales, menu_prices, members)
    
    print(f"\n📍 WESTSHORE LOCATION")
    print(f"Revenue: ${westshore_rev:.2f}")
    print(f"Transactions: {len(westshore_sales)}")
    print(f"Average Order: ${analytics.calculate_average_order(westshore_sales):.2f}")
    
    print(f"\n📍 DOWNTOWN LOCATION")
    print(f"Revenue: ${downtown_rev:.2f}")
    print(f"Transactions: {len(downtown_sales)}")
    print(f"Average Order: ${analytics.calculate_average_order(downtown_sales):.2f}")
    
    # Compare locations
    print(f"\n📊 LOCATION COMPARISON")
    comparison = reports.compare_locations(westshore_sales, downtown_sales, 
                                          metric='revenue')
    for loc, revenue in comparison.items():
        print(f"{loc}: ${revenue:.2f}")
    
    # Best sellers
    print(f"\n🏆 BEST SELLERS (Combined)")
    all_sales = westshore_sales + downtown_sales
    best_sellers = analytics.find_best_sellers(all_sales, 5)
    for rank, (item, count) in enumerate(best_sellers, 1):
        print(f"#{rank}: {item} ({count} sold)")
    
    # Summary
    print(f"\n📈 SUMMARY")
    total_revenue = westshore_rev + downtown_rev
    total_transactions = len(westshore_sales) + len(downtown_sales)
    print(f"Total Daily Revenue: ${total_revenue:.2f}")
    print(f"Total Transactions: {total_transactions}")
    print(f"Overall Average Order: ${total_revenue/total_transactions:.2f}")

# Run the main program
if __name__ == "__main__":
    main()
````

## 💡 Tips for Success

### Common Issues and Solutions:

1. **Module not found error**: Make sure you run the `%%writefile` cells before importing

2. **Function returns None**: Make sure you have a `return` statement

3. **KeyError with dictionary**: Check if key exists before accessing:
   ```python
   if item in prices:
       total += prices[item]
   ```

4. **Division by zero**: Check for empty lists:
   ```python
   if len(sales_list) > 0:
       average = total / len(sales_list)
   ```

### Debugging Tips:

- Add print statements to see what's happening:
  ```python
  print(f"Debug: total = {total}")
  ```

- Test with simple data first:
  ```python

  ```
